#**B. Analytical Approach**

**Variabel Target:**
- 0 : calon pengunjung tidak membatalkan ruang yang sudah dipesan
- 1 : calon pengunjung membatalkan ruang yang sudah dipesan

**Gambaran umum biaya dalam bisnis hotel dan kegiatan hospitality:**

- Rata-rata upah minimum per bulan bagi pekerja dalam bidang hotelier/hospitality di Portugal periode 2014-2015 adalah 935 euro; 11.120 euro per tahun. [Sumber](https://www.reinisfischer.com/average-salary-european-union-2015)
- Rata-rata biaya sewa hotel bintang empat di Portugal periode 2014-2015 adalah 187 euro per malamnya [Sumber](https://www.kayak.com/Portugal-Hotels.197.dc.html#:~:text=The%20average%20price%20of%20a,with%20prices%20varying%20by%20date.)
- Rata-rata biaya untuk menjalankan 1 kamar, beserta dengan karyawan, dan fasilitas yang mendukung keberlangsungan hotel di Portugal berada di angka 120 euro per malamnya [Sumber](https://www.researchgate.net/publication/264563014_Cost_Efficiency_of_Portuguese_Hotels_in_the_Algarve_A_Comparative_Analysis_Using_Mathematical_and_Econometric_Approaches)

**Evaluation Metrics:**

- _True Positive (TP)_ : Pengunjung Hotel yang diprediki akan datang dan benar datang bertamu ke hotel.
- _False Positive (FP)_ : Pengunjung yang diprediksi akan datang, namun tidak datang bertamu.
- _True Negative (TN)_ : Pengunjung yang diprediksi tidak akan datang, dan tidak datang.
- _False Negative (FN)_ : Pengunjung yang diprediksi tidak datang, namun ternyata datang.


- _**Recall**_ : Definisi: Mengukur seberapa akurat prediksi model saat menyatakan bahwa seorang pengunjung akan datang. Dari semua pengunjung yang diprediksi datang oleh sistem, berapa persen yang benar-benar datang?.
  - **Rumus Recall** = TP / (TP + FN)
  _Untuk meminimalkan False Negative_

- _**Precision**_ : Mengukur seberapa akurat prediksi model saat menyatakan bahwa seorang pengunjung akan datang. Dari semua pengunjung yang diprediksi datang oleh sistem, berapa persen yang benar-benar datang?
  - **Rumus Precision** = TP / (TP + FP)
  _Untuk meminimalkan False Positive_

- **F1-Score** : Nilai rata-rata harmonis (harmonic mean) dari Precision dan Recall. Metrik ini digunakan untuk mencari keseimbangan antara keduanya.
  - **Rumus F1-Score** = 2 * (Recall * Precision) / (Recall + Precision)
  _untuk mencari keseimbangan antara FP dan FN_

####**uji normalitas**

```python
def check_normality(df):
    def test_normality(df):
        shapiro_p = shapiro(df).pvalue
        ks_p = kstest(df, 'norm').pvalue
        anderson_result = anderson(df)
        ad_result = anderson_result.statistic < anderson_result.critical_values[2] #using 5% critical value

        return {
            'Shapiro-Wilk': 'Normal' if shapiro_p > 0.05 else 'Not-normal',
            'Kolmogorov-Smirnov': 'Normal' if ks_p > 0.05 else 'Not-normal',
            'Anderson-Darling': 'Normal' if ad_result else 'Not-normal'
        }

    numeric_cols = df.select_dtypes(include=["number"]).columns
    results = {col: test_normality(df[col]) for col in numeric_cols}
    return pd.DataFrame.from_dict(results, orient='index').rename_axis('Feature')

normality_results = check_normality(df)
normality_results
```

```python
#CELL 18 Menambah Fitur Baru
df_prep = df.copy()

# 1. Total nights stay
if {'stays_in_weekend_nights','stays_in_week_nights'}.issubset(df_prep.columns):
    df_prep['total_nights'] = df_prep['stays_in_weekend_nights'] + df_prep['stays_in_week_nights']
    print("   ✅ Created 'total_nights'")

# 2. Total people
people_cols = ['adults','children','babies']
existing_people_cols = [c for c in people_cols if c in df_prep.columns]
if existing_people_cols:
    df_prep['total_people'] = df_prep[existing_people_cols].sum(axis=1)
    print("   ✅ Created 'total_people'")

# 3. Weekend ratio
if {'stays_in_weekend_nights','total_nights'}.issubset(df_prep.columns):
    df_prep['weekend_ratio'] = df_prep['stays_in_weekend_nights'] / (df_prep['total_nights'] + 1e-8)
    print("   ✅ Created 'weekend_ratio'")


# 4. Booking complexity
complexity_features = ['booking_changes','total_of_special_requests']
existing_complexity = [c for c in complexity_features if c in df_prep.columns]
if existing_complexity:
    df_prep['booking_complexity'] = df_prep[existing_complexity].sum(axis=1)
    print("   ✅ Created 'booking_complexity'")

# 5. Customer reliability
if {'previous_cancellations','previous_bookings_not_canceled'}.issubset(df_prep.columns):
    total_prev = df_prep['previous_cancellations'] + df_prep['previous_bookings_not_canceled']
    df_prep['customer_reliability'] = df_prep['previous_bookings_not_canceled'] / (total_prev + 1e-8)
    print("   ✅ Created 'customer_reliability'")

# 6. Lead time categories
if 'lead_time' in df_prep.columns:
    df_prep['lead_time_category'] = pd.cut(
        df_prep['lead_time'],
        bins=[-1, 7, 30, 90, 365, df_prep['lead_time'].max()],
        labels=['Very_Short','Short','Medium','Long','Very_Long']
    )
    print("   ✅ Created 'lead_time_category'")

# 7. Seasonal indicators
if 'arrival_date_month' in df_prep.columns:
    season_map = {
        'December':'Winter','January':'Winter','February':'Winter',
        'March':'Spring','April':'Spring','May':'Spring',
        'June':'Summer','July':'Summer','August':'Summer',
        'September':'Fall','October':'Fall','November':'Fall'
    }
    df_prep['season'] = df_prep['arrival_date_month'].map(season_map)
    peak_months = ['June','July','August','December']
    df_prep['is_peak_season'] = df_prep['arrival_date_month'].isin(peak_months).astype(int)
    print("   ✅ Created 'season' and 'is_peak_season'")
```

##### Fitur Tambahan

| Feature                | Definisi                                                                 | Tujuan                                                                                     | Manfaat                                                                 |
|------------------------|--------------------------------------------------------------------------|--------------------------------------------------------------------------------------------|-------------------------------------------------------------------------|
| **total_nights**       | `stays_in_weekend_nights + stays_in_week_nights`                        | Menyederhanakan lama menginap tamu                                                         | Indikator komitmen tamu (lama stay → jarang cancel)                     |
| **total_people**       | `adults + children + babies`                                            | Menggambarkan ukuran rombongan tamu                                                        | Reservasi dengan banyak orang lebih serius, bisa untuk segmentasi       |
| **weekend_ratio**      | `stays_in_weekend_nights / total_nights`                               | Mengukur preferensi weekend stay                                                           | Bedakan leisure (weekend) vs business (weekday) traveler                |
| **booking_complexity** | `booking_changes + total_of_special_requests`                          | Menggambarkan kompleksitas reservasi                                                       | Banyak request = lebih serius, tapi banyak perubahan = tanda ketidakpastian |
| **customer_reliability** | `previous_bookings_not_canceled / (previous_cancellations + previous_bookings_not_canceled)` | Mengukur track record tamu                                                                 | Reliability tinggi → lebih dipercaya, cancel lebih rendah               |
| **lead_time_category** | Kategori dari `lead_time` (Very_Short, Short, Medium, Long, Very_Long) | Mengelompokkan perilaku booking (last minute vs early booking)                             | Early booking biasanya punya peluang cancel lebih besar                  |
| **is_peak_season**     | Indikator apakah booking pada musim ramai (contoh: Jun–Aug, Dec)        | Mengukur dampak musim terhadap pola reservasi                                              | Peak season → cancel rate lebih rendah, relevan untuk revenue management |
| **season**             | Kategori musim dari `arrival_date_month` (Winter, Spring, Summer, Fall) | Mengelompokkan pola musiman booking                                                        | Berguna untuk analisis tren, strategi pricing, dan segmentasi pasar     |
